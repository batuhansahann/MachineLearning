- **Method:** Data points are clustered based on their proximity to a "centroid."
- **Algorithms:**  
  - **K-Means:** Creates K clusters and assigns a center to each cluster.  
  - **K-Means++:** Improves the K-Means algorithm by selecting better initial centroids.  
  - **K-Medoids:** Uses actual data points within the cluster as centroids instead of calculated centers.

- **Method:** Based on the distance or connection between data points. Points are clustered according to their proximity to each other.
- **Algorithms:**  
  - **Hierarchical Clustering:**  
    - **Agglomerative:** Each data point starts as its own cluster; similar points merge to form clusters.  
    - **Divisive:** All data points start in a single cluster, which is then divided into subgroups.

- **Method:** Data points are clustered based on their density. Regions with higher densities are treated as clusters.

---

**Graph-Based Clustering**
- **Method:** Represents the distances between data points as a graph and forms clusters based on the connections in the graph.

# <font color='orange'> <b>K-Means Clustering</b><font color='orange'>  

**1. Introduction to K-Means Clustering**

K-Means clustering is a popular technique among **unsupervised learning** methods. It divides data into a predefined number of \( K \) clusters and assigns each data point to the nearest cluster center (**centroid**). The algorithm aims to reveal hidden patterns and structures by grouping data points with similar characteristics within clusters.

**The Core Concept of K-Means:**
- **Objective:** Minimize the total **variance** within clusters by assigning each data point to the nearest cluster center.
- **Optimization:** The algorithm minimizes the sum of squared distances between data points and their cluster centers (**within-cluster sum of squares, WCSS**).

---

**2. Why K-Means is Important**

The K-Means algorithm is crucial for making sense of unlabeled data in large datasets. It is widely applicable across various fields and offers the following advantages:

1. **Simplicity and Speed:** K-Means is fast and easy to implement.  
2. **Versatility:** Applicable in various domains, such as:  
   - Marketing (customer segmentation)  
   - Image processing (image segmentation)  
   - Text analysis (document clustering)  
3. **Scalability:** Suitable for handling large datasets.

---

**3. Understanding Clustering Algorithms**

Clustering is an **unsupervised learning** technique used to group similar data points. K-Means is one of the most widely used clustering algorithms, but other methods may offer different advantages:

**Types of Clustering Algorithms:**

1. **Hierarchical Clustering:**  
   Builds a hierarchy among clusters (e.g., agglomerative and divisive methods).  
   
2. **Partition-Based Clustering:**  
   Algorithms like K-Means divide the dataset into a predefined number of \( K \) groups.  
   
3. **Density-Based Clustering:**  
   Algorithms like DBSCAN use a density-based approach to detect clusters, even those with irregular shapes.

---

**4. How K-Means Clustering Works**

The K-Means algorithm follows these steps:

1. **Determine the Number of Clusters (\( K \)):**

   - Decide the number of clusters to form.

2. **Initialize Centroids:**

   - Select \( K \) random initial points as centroids, which are treated as the initial cluster centers.
   - Centroids are randomly chosen (or strategically selected using K-Means++ for better results).

3. **Assign Points to Clusters:**

   - Assign each data point to the nearest centroid based on the calculated distance.
   - Distance is typically calculated using **Euclidean Distance**.

4. **Update Centroids:**

   - Update each cluster's centroid by calculating the mean of the data points within that cluster.

5. **Convergence:**

   - The algorithm stops when data point assignments do not change, or the centroids stabilize.

---

**5. Applications of K-Means Clustering**

Some of the areas where K-Means is applied include:

1. **Customer Segmentation:**  
   Grouping customers based on purchase behavior, demographics, or other characteristics to tailor marketing strategies.

2. **Image Segmentation:**  
   Dividing an image into different regions (e.g., based on color or texture) to facilitate object recognition or image compression.

3. **Document Clustering:**  
   Grouping texts by topic, content, or sentiment to simplify information management.

4. **Anomaly Detection:**  
   Identifying outliers or unusual behaviors (e.g., detecting abnormal activities in network security).

---

**6. Choosing the Right Number of Clusters**

Determining the appropriate value for \( K \) in the K-Means algorithm is essential. The following methods can help:

1. **Elbow Method:**  
   Calculate WCSS (within-cluster sum of squares) for different values of \( K \) and plot the results. The "elbow" point in the graph typically indicates the ideal \( K \).

2. **Silhouette Score:**  
   Compares the similarity of a data point to its own cluster with its dissimilarity to other clusters. Higher scores indicate better-separated clusters.

---

**7. Challenges with K-Means Clustering**

1. **Sensitivity to Initialization:**  
   Randomly initialized centroids can lead to poor results. **K-Means++** addresses this issue by improving centroid initialization.

2. **Requires Predefined \( K \):**  
   The number of clusters must be specified beforehand, requiring prior knowledge of the dataset.

3. **Sensitive to Outliers:**  
   Outliers can skew the centroids, leading to inaccurate results.

4. **Assumes Spherical Clusters:**  
   K-Means is suitable for spherical and equally sized clusters but may fail with complex or irregularly shaped clusters.

---

**8. Advanced Variants of K-Means**

1. **K-Means++:**  
   Produces faster and more consistent results by strategically selecting centroids during initialization.

2. **Mini-Batch K-Means:**  
   Processes data in small batches to reduce computation time for large datasets.

3. **Fuzzy C-Means:**  
   Assigns data points to clusters based on probabilities, allowing partial membership.

---

**9. Conclusion**

K-Means is a foundational **clustering algorithm** in data science and machine learning. Due to its simplicity and efficiency, it is widely used for both small and large-scale datasets. However, its limitations (e.g., sensitivity to outliers and spherical cluster assumptions) necessitate evaluating alternative algorithms when needed. **K-Means++** and **Mini-Batch K-Means** provide practical solutions to some of these challenges.

**FAQs on K-Means Clustering**

1. **Where is K-Means clustering used?**  
   K-Means is used to group data by similarity. Common applications include customer segmentation, image segmentation, document clustering, and anomaly detection.

2. **How can I determine the optimal number of clusters in K-Means?**  
   Methods like the **Elbow Method**, **Silhouette Score**, and **Gap Statistics** can help determine the optimal number of clusters (\( K \)) for a dataset.

3. **Is K-Means sensitive to outliers?**  
   Yes, K-Means is sensitive to outliers. Outliers can distort centroids and lead to incorrect results. Preprocessing steps like **outlier removal** or **scaling** can mitigate this issue.

4. **Can K-Means handle large datasets?**  
   Yes, K-Means is a scalable algorithm and works efficiently with large datasets. Variants like **Mini-Batch K-Means** are specifically designed for large-scale data.

5. **What is K-Means++?**  
   K-Means++ is an improved initialization technique for centroids, leading to faster and more accurate clustering results.

6. **How does K-Means differ from hierarchical clustering?**  
   K-Means is a flat clustering method that directly divides data into a predefined number of clusters. Hierarchical clustering builds a **tree structure (dendrogram)**, allowing for more flexible exploration of clusters.

7. **Does K-Means assume a specific cluster shape?**  
   Yes, K-Means assumes spherical clusters of roughly equal size. For irregularly shaped or differently sized clusters, algorithms like **DBSCAN** or **Gaussian Mixture Models (GMM)** may perform better.

8. **What happens if an incorrect number of clusters (\( K \)) is chosen?**  
   Too few clusters may merge distinct groups, while too many clusters can overfit the data. Testing different \( K \) values and evaluating with metrics like the **Silhouette Score** can help identify the optimal value.

9. **How can high-dimensional data clustered with K-Means be visualized?**  
   High-dimensional data can be reduced to 2D or 3D using techniques like **Principal Component Analysis (PCA)** or **t-SNE**, and distributions can be visualized using scatter plots.

10. **What are the limitations of the K-Means algorithm?**  
    - **Predefined \( K \):** The number of clusters must be known beforehand.  
    - **Initialization Sensitivity:** Poor initial centroid selection can lead to suboptimal results.  
    - **Cluster Shape Assumption:** Not suitable for irregularly shaped clusters.  
    - **Outlier Sensitivity:** Outliers can significantly distort results.

# <font color='orange'> <b>Hopkins Test</b><font color='orange'>  

**What is the Hopkins Test?**

The Hopkins test is a statistical test used to measure the **clustering tendency** of a dataset. It helps determine how well the dataset is suited for clustering or whether the data points are randomly distributed. Performing this analysis before applying clustering algorithms like K-Means is crucial for assessing the algorithm's effectiveness.

The Hopkins test produces a value between 0 and 1:  
- **Below 0.5:** The dataset is suitable for clustering.  
- **0.3 and Below:** Indicates a strong clustering tendency (desired scenario).  
- **0.5 and Above:** The dataset is randomly distributed, and distinct clusters are unlikely.

---

**Step-by-Step Implementation of the Hopkins Test**

1. **Sampling:**  
   - A random group of data points is selected from the dataset. This sample forms the basis of the test.

2. **Nearest Neighbor Distances for Real Points:**  
   - For each selected real data point, the distances to the nearest neighbors within the dataset are calculated.

3. **Random Points Generation:**  
   - A group of completely random points is generated within the range and dimensions of the dataset.

4. **Nearest Neighbor Distances for Random Points:**  
   - For these randomly generated points, the distances to the nearest neighbors in the dataset are calculated.

5. **Calculating the Hopkins Score:**  
   - The total nearest neighbor distances of the real data points (W) are compared to those of the random points (U).  
   


---

**Interpretation of Results**

- **\( H \) < 0.5:** The dataset is suitable for clustering algorithms.  
- **\( H \) \(\approx\) 0.3 or lower:** Indicates a strong clustering tendency (desired scenario).  
- **\( H \) ≥ 0.5:** The dataset is randomly distributed, with no distinct clustering structure.

---

**Why is the Hopkins Test Important?**  
The Hopkins test is a quick and effective method to understand whether your dataset is clusterable. The test result helps evaluate the applicability of algorithms like K-Means, saving time and resources by avoiding unsuitable datasets.

````python
# pip install pyclustertend
# from pyclustertend import hopkins
````

````python
from sklearn.neighbors import BallTree  # BallTree is a data structure used to quickly perform distance queries between data points.
import numpy as np
import pandas as pd

def hopkins(data_frame, sampling_size):
    
    """Assess the clusterability of a dataset. A score between 0 and 1, a score around 0.5 express
    no clusterability and a score tending to 0 express a high cluster tendency.
    Parameters
    ----------
    data_frame : numpy array
        The input dataset
    sampling_size : int
        The sampling size which is used to evaluate the number of DataFrame.
    Returns
    ---------------------
    score : float
        The hopkins score of the dataset (between 0 and 1)
    Examples
    --------
    >>> from sklearn import datasets
    >>> from pyclustertend import hopkins
    >>> X = datasets.load_iris().data
    >>> hopkins(X,150)
    0.16
    """
    
    if type(data_frame) == np.ndarray:
        data_frame = pd.DataFrame(data_frame)
    
    # Sample n observations from D : P
    if sampling_size > data_frame.shape[0]:
        raise Exception(
            'The number of sample of sample is bigger than the shape of D')
    data_frame_sample = data_frame.sample(n=sampling_size)
    
    # Get the distance to their neirest neighbors in D : X
    tree = BallTree(data_frame, leaf_size=2)
    dist, _ = tree.query(data_frame_sample, k=2)
    data_frame_sample_distances_to_nearest_neighbours = dist[:, 1]
    
    # Randomly simulate n points with the same variation as in D : Q.
    max_data_frame = data_frame.max()
    min_data_frame = data_frame.min()
    uniformly_selected_values_0 = np.random.uniform(min_data_frame[0], max_data_frame[0], sampling_size)
    uniformly_selected_values_1 = np.random.uniform(min_data_frame[1], max_data_frame[1], sampling_size)
    uniformly_selected_observations = np.column_stack((uniformly_selected_values_0, uniformly_selected_values_1))
    if len(max_data_frame) >= 2:
        for i in range(2, len(max_data_frame)):
            uniformly_selected_values_i = np.random.uniform(min_data_frame[i], max_data_frame[i], sampling_size)
            to_stack = (uniformly_selected_observations, uniformly_selected_values_i)
            uniformly_selected_observations = np.column_stack(to_stack)
    uniformly_selected_observations_df = pd.DataFrame(uniformly_selected_observations)
    
    # Get the distance to their neirest neighbors in D : Y
    tree = BallTree(data_frame, leaf_size=2)
    dist, _ = tree.query(uniformly_selected_observations_df, k=1)
    uniformly_df_distances_to_nearest_neighbours = dist
    
    # return the hopkins score
    x = sum(data_frame_sample_distances_to_nearest_neighbours)
    y = sum(uniformly_df_distances_to_nearest_neighbours)
    if x + y == 0:
        raise Exception('The denominator of the hopkins statistics is null')
    return x / (x + y)[0]
````

## <font color='orange'> <b>Elbow Method</b><font color='orange'>  

````python
ssd = []    # inertia  .. SSD, "Sum of Squared Distances"

K = range(2,10)  # starting with a minimum of 2 clusters

for k in K:                                             # Creates and trains the K-Means model for each different number of clusters.
    model = KMeans(n_clusters =k, random_state=42)
    model.fit(X)
    ssd.append(model.inertia_)

plt.plot(K, ssd, "bo-")
plt.xlabel("Different k values")
plt.ylabel("inertia-error")
plt.title("elbow method")
plt.show()
````

# <font color='orange'> <b>Silhouette Analysis</b><font color='orange'>  

**What is Silhouette Analysis?**

**Silhouette analysis** is a metric used to evaluate the performance of clustering. It compares intra-cluster similarity (**cohesion**) with inter-cluster similarity (**separation**) to show how well clusters are separated and whether data points are correctly clustered. A **silhouette value** is calculated for each data point, and the average of all values provides an overall measure of clustering quality.

The **Silhouette Score** is typically calculated using **Minkowski distance** or **Euclidean distance** and results in a value between **[-1, 1]**.

---

**Silhouette Value Ranges and Interpretations**

- **1:** The data point is very well-placed within its cluster and completely isolated from other clusters. (Good clustering)  
- **0:** The data point is near the boundary between two clusters. (Ambiguous clustering)  
- **-1:** The data point is misclassified and likely closer to another cluster. (Poor clustering)

---

**Practical Interpretation of Silhouette Values**  
While no strict thresholds are defined for silhouette values, the following general interpretations can be made:
- **Above 0.5:** Indicates good clustering performance.  
- **Between 0.25 and 0.5:** Clustering results could be improved.  
- **Below 0.25:** The clustering algorithm may not work well with the data.

---

**Calculating the Silhouette Score**  
The silhouette score is based on two key measurements:

1. **Cohesion:**  
   - The average distance between a data point and other points in the same cluster.  
   - A **low cohesion value** indicates the point is close to other points in its cluster.

2. **Separation:**  
   - The minimum average distance between a data point and points in another cluster.  
   - A **high separation value** shows that the point is far from other clusters.

Formula:



- \(a(i)\): The average distance between the point and other points in its own cluster (**cohesion**).  
- \(b(i)\): The average distance between the point and the nearest neighboring cluster (**separation**).

---

**Applications of Silhouette Analysis**

1. **Evaluating Clustering Quality:**  
   - Used to measure the success of your clustering algorithm.  

2. **Determining the Optimal Number of Clusters:**  
   - Calculate the silhouette score for different numbers of clusters and choose the \(k\) that provides the highest average score.  

3. **Improving the Clustering Model:**  
   - A low silhouette score indicates the need for improvements in the clustering algorithm or data preprocessing steps.  

**Silhouette analysis** is an effective way to evaluate the results of clustering algorithms. Using measures like **cohesion** and **separation**, the quality of cluster separation can be easily analyzed. By interpreting the scores, important insights can be obtained to improve the algorithm's performance.

---

**Two Key Criteria for Successful Clustering Algorithms**

1. **Close Proximity of Elements Within a Cluster:**  
   - The distances between elements within a cluster should be **minimized**.  
   - This represents intra-cluster consistency (**cohesion**) and enhances similarity among cluster members.  

2. **Maximum Distance Between Clusters:**  
   - The distances between the centers of different clusters should be **maximized**.  
   - This represents inter-cluster separation (**separation**) and ensures that clusters are clearly distinguished.  

These two criteria are fundamental for **improving clustering performance** and **achieving meaningful results**.

## <font color='orange'> <b>Adjust Rand Score</b><font color='black'>  

**What is the Rand Index?**

The **Rand Index** is a metric used to evaluate the performance of a clustering algorithm. Specifically, it measures how similar two different clustering results are—typically one produced by an algorithm's predictions and the other being the true labels. This metric is ideal for assessing the accuracy of clustering results based on correct classifications.

---

**Rand Index Value**

The Rand Index takes a value between **0 and 1**:
- **1:** Indicates a perfect match; the predicted clusters align completely with the true labels.  
- **0:** Indicates a random match; there is no alignment between the predicted clusters and the true labels.

---

**Interpretation of Rand Index Values**

| **Rand Index Range**    | **Interpretation**      |
|--------------------------|-------------------------|
| **0.1 and below**        | No agreement, weak match |
| **0.1 - 0.3**            | Low agreement           |
| **0.3 - 0.5**            | Moderate agreement      |
| **0.5 - 1.0**            | Good agreement, meaningful results |

---

**Rand Index Calculation**

The Rand Index is calculated as the ratio of correctly classified pairs to the total number of pairs. Correct classifications include:

1. **True Positive (TP):** Pairs correctly classified as belonging to the same cluster.  
2. **True Negative (TN):** Pairs correctly classified as belonging to different clusters.  

Formula:



---

**Rand Index Interpretation**

- **High Rand Index (0.5 - 1.0):** The clustering algorithm groups the dataset in a way that closely matches the true labels.  
- **Low Rand Index (0.1 - 0.3):** The algorithm clusters the dataset in a way that diverges significantly from the true labels.  

**The Rand Index is a powerful metric for evaluating the performance of unsupervised learning algorithms, especially when comparing clustering results to ground truth labels.**

---

**Adjusted Rand Index Calculation**

ARI is based on the Rand Index, which calculates the ratio of correct classifications to the total number of sample pairs, but it corrects for random matches, providing a more accurate metric.

Formula:



- **RI:** Rand Index value.  
- **Expected RI:** Expected Rand Index value for a random clustering result.  
- **Max RI:** Maximum Rand Index value (1).

This correction prevents random results from receiving a high score, making ARI a more reliable measure.

---

**Advantages of ARI**

1. **Elimination of Random Matches:** Random results are not awarded high scores.  
2. **Negative Scores:** Clusterings that perform worse than random matching receive negative scores.  
3. **More Accurate Evaluation:** Provides a more meaningful measure than Rand Index.

---

- **High ARI (0.5 - 1.0):** The clustering algorithm groups the dataset in a way that closely matches the true labels.  
- **Low ARI (0 - 0.25):** The algorithm clusters the dataset in a way that diverges significantly from the true labels.  
- **Negative ARI:** The clustering result is worse than random matching.

**Adjusted Rand Index** is a powerful and reliable metric for evaluating the performance of unsupervised learning algorithms. The adjustment for random matches makes ARI superior to the Rand Index.

---

| **Feature**                  | **Rand Index (RI)**                                   | **Adjusted Rand Index (ARI)**                                |
|------------------------------|-----------------------------------------------------|-------------------------------------------------------------|
| **Random Match Correction**   | Does not account for random matches, no correction.  | Accounts for random matches and adjusts the result accordingly. |
| **Value Range**               | Between 0 and 1                                      | Between -1 and 1                                            |
| **Random Clustering Score**   | Can take a positive value.                           | Usually around 0 or negative.                               |
| **Negative Values**           | Does not take negative values.                       | Can take negative values for worse-than-random results.     |
| **Sensitivity**               | Low (Sensitive to random matches).                   | High (Provides a more reliable evaluation).                 |
| **Use Case**                  | Suitable for simple performance measurements.        | Preferred for more accurate analysis as it penalizes random results. |
