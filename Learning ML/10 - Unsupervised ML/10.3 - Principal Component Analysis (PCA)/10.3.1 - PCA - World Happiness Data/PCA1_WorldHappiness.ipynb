**PCA (Principal Component Analysis)**

**Principal Component Analysis (PCA)** is a **dimensionality reduction** and **complexity reduction** technique.

PCA, especially in high-dimensional datasets, creates fewer new components (principal components) while retaining the main information, thus facilitating data analysis and visualization.

---

**General Features of PCA**
1. **Dimensionality Reduction:**
   - It creates **new components** from the existing features.
   - These components explain the variance in the dataset in the best possible way.
   - The goal is to represent most of the information in the dataset with fewer components.

2. **Feature Extraction:**  
   - PCA extracts new components from the existing features (**Feature Extraction**) but does not completely eliminate features (for example, it does not perform **Feature Selection** like **Lasso Regression**).

3. **Variance-Based:**
   - PCA creates components that **maximize the variance** in the dataset.
   - For instance, an explained variance of 80% is often considered a threshold.

4. **Solves Multicollinearity Problem:**
   - The components created by PCA are independent of each other, meaning there is **no multicollinearity** (no linear dependence between components).

5. **Supervised and Unsupervised Use:**
   - PCA can be used in both **supervised learning** (e.g., before regression or classification models) and **unsupervised learning** (e.g., clustering).

6. **Scaling Is Required:**
   - The features in the dataset need to be **scaled** before applying PCA.
   - This is because PCA can be affected by features with different scales when measuring variance.

---

**Explanation of PCA Concept with Examples**

1. **Pomegranate Example:**
   - PCA condenses the information in a dataset.  
   - For example: It's like squeezing 40 kg of pomegranates to get 2 kg of pomegranate juice. You extract the vitamins (principal components), but most of the mass is left behind.

2. **Flower Bouquet Example:**
   - PCA is like summarizing the general characteristics of a flower bouquet (such as color, size) instead of distinguishing between different types of flowers.

3. **Economic Example:**
   - It’s like creating a single indicator, such as GDP per capita, instead of using hundreds of economic variables like industry, agriculture, and tourism.

---

**Advantages of PCA**

1. **Dimensionality Reduction:**
   - **Data visualization** becomes easier. By reducing dimensions, it becomes possible to visualize high-dimensional datasets in 2D or 3D.

2. **Facilitates EDA (Exploratory Data Analysis):**
   - It is easier to understand key information with fewer components.

3. **Reduces Computation Time:**
   - Algorithms run faster because there are fewer features to process.

4. **Reduces Storage Space and Complexity:**
   - Working with less data requires less storage space.

5. **Reduces Overfitting Risk:**
   - By removing unnecessary features, the model complexity is reduced, and the risk of overfitting is decreased.

---

**Disadvantages of PCA**

1. **Data Loss:**
   - There may be some information loss during dimensionality reduction. However, this loss is typically acceptable (usually 10% or less).

2. **Difficulty in Interpreting Features:**
   - The components created may be hard to interpret since they are derived from the original features.

3. **Only Applicable to Numerical Data:**
   - PCA can only be applied to numerical data. Categorical data must be transformed before PCA.

4. **Scaling Requirement:**
   - PCA does not work properly with data that has different scales. Standardizing the data (e.g., using **StandardScaler**) is necessary.

---

**How PCA Works**

1. **Covariance Matrix:**  
   - PCA creates a **covariance matrix** to explain how features change together.

2. **Eigenvalues and Eigenvectors:**  
   - The **eigenvalues** and **eigenvectors** of the covariance matrix are calculated.
   - **Eigenvalues**: Represent variance (larger eigenvalues represent components explaining more variance).
   - **Eigenvectors**: Represent the directions of the new components.

3. **Creation of New Components:**  
   - Components with the highest eigenvalues are selected, and new components are created.

4. **Explained Variance Ratio (%):**  
   - The explained variance ratio of the components is calculated, and the number of components is determined based on these ratios.

---

**Variance and Number of Components**
- **Explained Variance Ratio:**  
  - This expresses how well the new components explain the total variance in the dataset. 
  - For example: If the variance explained is 80% or more, this threshold is generally considered acceptable.  
  - Components that explain 80% variance can be selected, and the remaining components can be discarded.

- **Which Components Are Selected?**
  - Components that explain 80% variance are chosen.  
  - For example: In a dataset with 100 features, if PCA creates 10 components explaining 90% of the variance, the remaining 90 features can be discarded.

---

**Use Cases for PCA**
1. **Supervised Learning:**
   - Reducing the number of features to make regression or classification models more efficient.
   - For example: Reducing the risk of **overfitting** in a dataset with too many features.

2. **Unsupervised Learning:**
   - Reducing the dimensionality in clustering algorithms and creating more distinct clusters.
   - For example: Performing clustering after PCA in **K-Means Clustering**.

3. **Data Visualization:**
   - Visualizing high-dimensional datasets by reducing them to 2D or 3D.

4. **Genetic and Biological Data:**
   - PCA is used in genetic studies to summarize variation in gene expression.

5. **Economics and Finance:**
   - Used to summarize and analyze macroeconomic indicators.

---

**Summary Principles of PCA**

- **Does not completely eliminate any features.** It creates new components considering the variance of the existing features.
- **Categorical data cannot be used.** Categorical data must be transformed before PCA.
- **Scaling is mandatory.** Features with different scales affect the accuracy of PCA.

---

**Difference Between PCA and Feature Selection**

| **Criterion**            | **PCA (Principal Component Analysis)**                                    | **Feature Selection**                                      |
|--------------------------|---------------------------------------------------------------------------|------------------------------------------------------------|
| **Purpose**              | Dimensionality reduction, summarizing data.                              | Selecting the most important features for the model and eliminating the irrelevant ones. |
| **Approach**             | Creates new components from the features.                                | Selects important features from the original ones and excludes others. |
| **Output**               | **Abstract components** derived from linear combinations of features.    | A smaller subset of the original features.                 |
| **Interpretation**       | Components are difficult to interpret; abstractions are present.         | The selected features are clearly related to the original features. |
| **Dependency (Supervised/Unsupervised)** | **Unsupervised**: Does not look at labels.                           | **Supervised**: Works by looking at labels.               |
| **Scaling**              | Scaling of data is mandatory.                                            | Generally not mandatory (depends on the algorithm).        |
| **Loss**                 | Reduces dimensionality without losing variance (some information loss may occur). | Completely removes irrelevant features affecting the model. |
| **Data Type**            | Works only with **numerical data**.                                      | Can work with both numerical and categorical data.         |
| **Example Techniques**   | PCA, Kernel PCA.                                                         | Decision Tree, Random Forest, Lasso, Recursive Feature Elimination (RFE). |

---

**When to Use Which?**

| **Situation**                                              | **PCA Usage**                                        | **Feature Selection Usage**                              |
|------------------------------------------------------------|------------------------------------------------------|----------------------------------------------------------|
| Working with very high-dimensional datasets.               | Yes (e.g., 1000 features -> 10 components).         | No (can be costly with high-dimensional datasets).       |
| Model interpretability is not critical.                    | Yes (components may be hard to interpret).          | No (original features are preserved and interpretable).   |
| Model performance and explanation are important.           | No (original features are lost).                    | Yes (the features the model depends on are clearly visible). |
| Dimensionality reduction and visualization are needed.     | Yes (e.g., reducing the dataset to 2D/3D).          | No (focuses on explanation rather than visualization).    |

# <font color='orange'> <b>PCA Hyperparameters</b><font color='black'>  

**PCA with Sklearn and Prince**

The PCA algorithm available in the Scikit-learn library is a highly effective tool for data analysis. However, the **Prince** library is often preferred in machine learning applications due to its powerful **visualization tools**.

---

**Prince PCA Parameters and Their Meanings**

1. **n_components=6**:  
   - Specifies how many principal components the PCA will reduce the data to.  
   - In this case, the data will be reduced to 6 dimensions.

2. **n_iter=10**:  
   - Indicates the number of iterations the PCA algorithm will perform.  
   - More iterations can yield better results.

3. **rescale_with_mean=False**:  
   - Specifies whether the data should be rescaled based on the mean.  
   - False: The data will not be rescaled by the mean.

4. **rescale_with_std=False**:  
   - Specifies whether the data should be rescaled based on the standard deviation.  
   - False: The data will not be rescaled by the standard deviation.

5. **copy=True**:  
   - Indicates whether PCA will copy the data before processing it or directly modify the original data.  
   - True: The original data is not altered; a copy is used instead.

6. **check_input=True**:  
   - Specifies whether the input data will be validated.  
   - True: Data validation is performed before running the PCA algorithm.

7. **engine="sklearn"**:  
   - Specifies the engine to be used for PCA computation.  
   - For example, options like 'sklearn', 'fbpca', or 'auto' can be used.  
   - In this example, Scikit-learn is being used.

**1. What is an Eigenvector?**
- An eigenvector represents the **principal directions** or **main components** in your dataset.  
- Eigenvectors define the directions of the variables in the dataset, and each eigenvector represents one dimension of the dataset.  
- In PCA, eigenvectors form the **principal component axes**, which are newly defined axes in the data space that capture the maximum variance in the data.  

For example:  
If your data lies in a 2D plane, the eigenvectors represent the directions in which the data cloud spreads the most (where variance is maximized).

---

**2. What is an Eigenvalue?**
- An eigenvalue indicates the **magnitude of variance** along the direction represented by an eigenvector.  
- The larger the eigenvalue, the more information (variance) the eigenvector can explain.  
- In PCA, eigenvalues are ranked, and the eigenvector with the largest eigenvalue forms the first principal component.

For example:  
If one eigenvalue is 5 and the other is 1, the first eigenvector (first principal component) explains a significant portion of the variance, while the second eigenvector explains much less.

---

**PCA Process and Its Relationship**

1. **Compute the Covariance Matrix:** The relationships between the variables in the dataset are calculated.  
2. **Extract Eigenvectors and Eigenvalues:** Eigenvectors and eigenvalues are computed from the covariance matrix.  
   - **Eigenvectors:** Define the principal directions in the dataset.  
   - **Eigenvalues:** Indicate the amount of variance explained by each principal direction.  
3. **Select Components:** The eigenvectors with the largest eigenvalues are selected, and the data is projected onto these new axes.

---

- **Eigenvectors:** Represent the fundamental directions (principal components) of the data.  
- **Eigenvalues:** Indicate the amount of variance explained by each direction.

```python
import prince
pca = prince.PCA(n_components=6, n_iter=10, rescale_with_mean=False,
rescale_with_std=False, copy=True, check_input=True,
engine= "sklearn" , random_state=42)

# Entering values ​​into hyperporometers for PCA model
# There are already 6 components, let's take them all into consideration.
```

```python
pca = pca.fit(df2[features])
```

```python
# eigeanvalues of the model
```

```python
# Selection of the optimum number of basic components with the Elbow method

ev = pd.DataFrame()
ev["pca"] = range(1,7)
ev["eigenvalue"]=pd.DataFrame(pca.eigenvalues_)
plt.figure(figsize=(8,6))
sns.lineplot(x="pca", y = "eigenvalue", marker = "o", data= ev)
plt.ylabel("Eigenvalue", fontsize=16)
plt.xlabel("Basic Components", fontsize=16)
plt.show()
```

```python
# Explained Variance Ratios (cumulative)
# The answer to the question of what extent does which principal component explain the phenomenon under investigation?

explained_variance=pca.cumulative_percentage_of_variance_
print(explained_variance)
```

```python
# Explained Variance Ratios (cumulative)
# Rule of thumb: How many principal components explain at least 80% of the variance?

avo = pd.DataFrame()
avo["pca"] = range(1,7)
avo["vari"] = pd.DataFrame(pca.cumulative_percentage_of_variance_)

plt.figure(figsize=(8,6))
graph = sns.barplot(x="pca", y ="vari", data=avo)

for p in graph.patches :
    graph.annotate('{:.2f}'.format(p.get_height()),(p.get_x()+0.2,p.get_height()), ha="center", va="bottom", color="black")

plt.ylabel("Explained Variance Ratios (%)", fontsize=18)
plt.xlabel("Basic Components", fontsize=18)
plt.show()  
```

```python
correlations = pca.column_correlations
correlations
```

```python
# Plot of Positions of Variables in Principal Components 1 and 2

scatter = pca.column_correlations.reset_index()

plt.figure(figsize=(10,6))

ax = sns.scatterplot(x=0, y=1, data=scatter)
ax.set(ylim=(-1,1), xlim=(-1,0))

def label_point(x, y, val, ax):
    a = pd.concat({"x": x, "y": y, "val": pd.Series(val)}, axis=1)  # Using scatter.index as Series
    for i, point in a.iterrows():
        ax.text(point["x"]+.02, point["y"], str(point["val"]))

label_point(scatter[0], scatter[1], scatter.index, plt.gca())
plt.axvline(-0.5, ls="--")
plt.axhline(0, ls="--")
plt.title("Positions of Variables in Principal Components 1 and 2", fontsize=18)
plt.xlabel("Basic Components 1 (49%)", fontsize=18)
plt.ylabel("Basic Components 2 (24%)", fontsize=18)
plt.show()
```

# <font color='orange'> <b>Cluster Prediction</b><font color='black'>

```python
# Clustering predictions based on different pairs of components

pd.DataFrame.iteritems = pd.DataFrame.items
import plotly.express as px
from plotly.offline import plot
components = pca.fit_transform(df2[features])
labels = {str(i): f"PC {i+1} ({var:.1f}%)"
          for i, var in enumerate(pca.cumulative_percentage_of_variance_)}
fig = px.scatter_matrix(components, labels=labels, dimensions=range(4),
                        color=df["class"])
fig.update_traces(diagonal_visible=False)
plot(fig)
```
