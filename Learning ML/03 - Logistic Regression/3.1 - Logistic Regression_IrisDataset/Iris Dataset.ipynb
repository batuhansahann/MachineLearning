# Exploratory Data Analysis

**Dataset Overview**

The **Iris dataset** is a classic dataset frequently used in the fields of machine learning and statistics. It was first introduced by Ronald Fisher in 1936 and is commonly used to distinguish between three different species of Iris flowers (Iris setosa, Iris virginica, and Iris versicolor). This dataset is widely used for training and testing classification algorithms in machine learning and statistical applications.

**Details of the Iris Dataset:**

The Iris dataset contains a total of 150 flower samples, with each species of Iris flower represented by 50 samples:

- **Iris-setosa**
- **Iris-versicolor**
- **Iris-virginica**

For each flower sample, four features have been recorded:

1. **sepal_length**: The length of the sepal (the outer part of the flower that protects the petals).
2. **sepal_width**: The width of the sepal.
3. **petal_length**: The length of the petal (the colorful, showy part of the flower).
4. **petal_width**: The width of the petal.

These features are measured in centimeters (cm).

**Objective:**

The objective is to create a model that can predict the species of the flower using these four features. In other words, the Iris dataset is used as a dataset for a supervised learning problem. The goal of the model is to accurately classify the flower species based on the given features.

**Summary:**

The Iris dataset contains 150 flower samples, each representing one of three Iris species. Four features (sepal_length, sepal_width, petal_length, petal_width) have been measured for each sample. These features are used to distinguish between the different Iris species.

- **Sepal Length:** The length of the sepal.
- **Sepal Width:** The width of the sepal.
- **Petal Length:** The length of the petal.
- **Petal Width:** The width of the petal.

This dataset is an ideal example for evaluating and comparing the performance of classification algorithms.

# Train | Test Split and Scaling

**Information about the Stratify Parameter**

The **Stratify** parameter is a parameter used in functions such as **train_test_split** in the Scikit-Learn library to maintain the class distribution of the target variable when splitting the dataset into training and test sets. This is especially important in classification problems, when there is an imbalance between classes.

**Importance of Using Stratify**

When working with unbalanced data sets, randomly allocating data may not reflect the original distribution of classes in the training and test sets. This can lead to unreliable model performance and evaluation metrics. By using the Stratify parameter, classes are represented proportionally in both sets.

**Benefits of the Stratify Parameter**

1. **Preserves Class Distribution:** Preserves the proportions of classes in the training and test sets in the original data set.
2. **Improves Model Evaluation:** Provides a more accurate evaluation of the model's performance on unseen data.
3. **Prevents Bias:** Prevents the model from performing well in the majority class and poorly in the minority class.

Use of **Stratify Parameter**

The stratify parameter is set as the target variable (usually **y**). In this way, the class distribution of the target variable is preserved when separating the data into training and test sets. For example, if in the original dataset the classes are in the ratio of 70%-30%, the same ratio is preserved in the separation using stratify.

**Result**

The use of the stratify parameter helps to obtain more accurate and reliable models by preserving the distribution of classes when splitting data sets, especially in data sets containing unbalanced classes. This is one of the best practices for evaluating the performance of the model in classification problems.

# Modelling and Model Performance

**Metrics for Multiclass Problems**

1. **Macro Average (Macro Avg):** 
    - F1 score (or other metric) is calculated for each class.
    - These values are summed and divided by the number of classes to get the average.
    - Macro average represents situations where all classes are equally important.
    - In unbalanced data sets, the performance of classes with fewer samples is equally important.
    - You can use this metric especially when you want all classes to be evaluated equally.

2. **Weighted Average (Weighted Avg):** 
    - The F1 score (or other metric) is calculated for each class.
    - These values are weighted by the number of samples in each class and averaged.
    - Weighted average takes into account the importance of classes depending on the number of instances.
    - The performance of classes with many instances is weighted more heavily.
    
3. **Micro Average (Micro Avg):**     

    - A global F1 score is calculated by summing the number of true positives (TP), false positives (FP) and false negatives (FN) in all classes.
    - Metrics such as precision, recall and F1 score are calculated globally, not on a class basis.
    - It is calculated over the sum of predictions in all classes.
    - It gives more weight to the performance of the majority class.
    - Useful for measuring overall model performance, especially when the data set is balanced.

- If you think that the performance of all classes is equally important, you should look at the ‘Macro Avg’ metric.
- However, if you think that some classes have more instances and the performance of these classes is more critical, you should focus on the ‘Weighted Avg’ metric.

- **Micro Recall**: This method calculates a single aggregate metric by summing the predictions made over all classes. That is, it uses the sum of the true positive (TP) and false negative (FN) values of all classes. This shows how the model performs overall across all classes. It is mostly preferred when class imbalances are not significant.

- **Macro Recall**: Calculates recall for each class separately and then takes a simple average of these recall values. This approach treats all classes with equal importance and evaluates the performance of each class with equal weight. This method is preferable in cases of class imbalance because it does not ignore the performance of minority classes.

**Difference Between Micro and Macro in General**

- **Micro-level metrics** focus on the aggregated performance of the model across all classes. This shows the overall performance of the model but may ignore imbalances between classes.
  
- **Macro-level metrics** evaluate the performance of each class individually and average these values. This approach evaluates the performance of all classes with equal importance, so the performance of minority classes is as important as that of major classes.

**When Should I Use Which?

- The only time you should use micro-level metrics is when you want to measure overall model performance, especially when class imbalance is not important.

- You should use micro-level metrics when you want to evaluate the performance of all classes with equal importance, especially when you are working with imbalanced data sets where you care about the performance of a minority of classes.


**When Do We Look at Macro Scores, When Do We Look at Accuracy?

Unbalanced data sets are situations where some classes have many more instances than others. For example, in a model diagnosing a disease, the number of instances that do not have the disease (negative) may be much higher than the number of instances that have the disease (positive). In this case, there is a risk that the model learns the majority class (negative examples) and ignores the minority class (positive examples). It is important to understand which metrics should be used in such unbalanced data sets.

**Macro Metrics**

In unbalanced data sets, it is necessary to look at ‘macro’ metrics. Macro metrics simply average the performance metrics calculated separately for each class. For example, the ‘macro F1 score’ is the arithmetic mean of the F1 scores calculated separately for each class. This makes it difficult to ignore the performance of a minority class and allows the model to learn all classes in a balanced way.

**Features of Macro Metrics:**
- Evaluates the performance of each class separately.
- Shows whether the model learns all classes in a balanced way.
- Makes it difficult to ignore the performance of minority classes.

**Accuracy Metrics**

The Accuracy metric can often be misleading in unbalanced data sets. If a model consistently predicts the majority class correctly, it may achieve a high accuracy score. This does not necessarily mean that the model performs well on the minority class. Therefore, ‘macro’ metrics are preferred over accuracy in imbalanced data sets.

**Limitations of Accuracy Metrics:**
- May be misleading in imbalanced data sets.
- Even if the majority class is predicted correctly, the performance of the minority class may be ignored.
- It does not reflect the success of the model in learning minority classes.

**Conclusion**

Using macro metrics to evaluate model performance in imbalanced data sets provides a more accurate approach as it evaluates the performance of each class separately and does not ignore the performance of minority classes. The Accuracy metric should be used with caution as it can be misleading in imbalanced data sets. 

If a balanced performance of the model over all classes is desired, focusing on macro metrics would be the best approach. This ensures that the model performs well not only in the majority class but also in the minority classes, thus providing a more holistic view of the overall model performance.

**Iris Dataset and Confusion Matrix Analysis**

The Iris dataset is a classical dataset used for a multi-class classification problem. In this dataset, three different Iris flower species (Iris setosa, Iris versicolor, Iris virginica) are classified. The confusion matrix, which is used to evaluate the classification performance, is an important tool that shows how well the model distinguishes each class.

**Confusion Matrix Structure**

Confusion matrix is a 3x3 matrix and shows how each class is confused with other classes. Each row represents the actual class and each column represents the predicted class.

**Confusion Matrix Analysis for Iris versicolor**

- **True Positive (TP) for Iris versicolor**: The ability of the model to correctly classify samples that are indeed Iris versicolor as Iris versicolor. In the confusion matrix, the cell where the Iris versicolor row and column intersect indicates the number of TPs.
- **Iris versicolor for False Positive (FP)**: The model incorrectly classifies non-Iris versicolor specimens (i.e. Iris setosa or Iris virginica) as Iris versicolor. In the confusion matrix, the Iris versicolor column but not Iris versicolor rows show FP counts.
- **False Negative (FN) for Iris versicolor**: The model mistakenly assigns samples that are really Iris versicolor to another class. In the confusion matrix, the Iris versicolor row but not the Iris versicolor columns show FN counts.

```python
def eval_metric(model, X_train, y_train, X_test, y_test):
    y_train_pred = model.predict(X_train)
    y_pred = model.predict(X_test)
    
    print("Test_Set")
    print(confusion_matrix(y_test, y_pred))
    print(classification_report(y_test, y_pred))
    print()
    print("Train_Set")
    print(confusion_matrix(y_train, y_train_pred))
    print(classification_report(y_train, y_train_pred))
```

```python
eval_metric(pipe_model, X_train, y_train, X_test, y_test)
```

```python
from sklearn.metrics import precision_score, recall_score, f1_score, accuracy_score

precision_macro = precision_score(y_test, y_pred, average='macro')
recall_macro = recall_score(y_test, y_pred, average='macro')
f1_macro = f1_score(y_test, y_pred, average='macro')

precision_micro = precision_score(y_test, y_pred, average='micro')
recall_micro = recall_score(y_test, y_pred, average='micro')
f1_micro = f1_score(y_test, y_pred, average='micro')

precision_weighted = precision_score(y_test, y_pred, average='weighted')
recall_weighted = recall_score(y_test, y_pred, average='weighted')
f1_weighted = f1_score(y_test, y_pred, average='weighted')

accuracy = accuracy_score(y_test, y_pred)

print("Macro Average Metrics")
print(f"Precision: {precision_macro:.3f}, Recall: {recall_macro:.3f}, F1-score: {f1_macro:.3f}")
print("---")
print("Micro Average Metrics")
print(f"Precision: {precision_micro:.3f}, Recall: {recall_micro:.3f}, F1-score: {f1_micro:.3f}")
print("---")
print("Weighted Average Metrics")
print(f"Precision: {precision_weighted:.3f}, Recall: {recall_weighted:.3f}, F1-score: {f1_weighted:.3f}")
print("---")
print(f"Accuracy: {accuracy:.3f}")
```

**Logistic Regression Hyperparameters**

1. **Penalty (penalty):**
   - **Description:** Determines the type of regularisation.
   - **Options:** ‘l1’, ‘l2’, ‘elasticnet’, ‘none’
     - ‘l1’: Lasso regularisation
     - ‘l2’: Ridge regularisation (default)
     - ‘elasticnet’: Combination of L1 and L2
     - ‘none’: No regularisation applied
   - Usage:** Used to prevent overfitting. L1 creates sparse models and reduces some coefficients to zero. L2 reduces all coefficients but does not make them zero.

2. **Inverse of Regularisation Strength (C):**
   - **Definition:** Corresponds to the inverse of the term Regularisation.
   - **Use:** The smaller the value of C, the stronger the regularisation. The larger the value of C, the weaker the regularisation.

3. **Solver (solver):**
   - Description:** Algorithm used to solve the optimisation problem.
   - **Options:** ‘newton-cg’, ‘lbfgs’, ‘liblinear’, ‘sag’, ‘saga’
     - ‘newton-cg’, ‘lbfgs’, ‘sag’, ‘saga’: Works with L2 penalty.
     - ‘liblinear’: Suitable for small data sets, works with L1 and L2 penalty.
     - ‘saga’: Suitable for large data sets, works with L1, L2, and elasticnet penalty.
   - **Usage:** The solver is selected according to the size of the data set and the type of regularisation.

4. **Maximum Number of Iterations (max_iter):**
   - Definition:** Maximum number of iterations of the optimisation algorithm.
   - Usage:** Determines the maximum number of iterations allowed for the optimisation algorithm to achieve convergence. In some cases, higher values provide better performance.

5. **Class Weight (class_weight):**
   - Description:** Weighting for classes.
   - **Options:** ‘balanced’, {dict, ‘balanced’}, None
     - ‘balanced’: Class weights are automatically adjusted, correcting class imbalances.
     - {dict}: Manually set weights for each class.
     - None: No weighting (default).
   - **Use:** Used to handle class imbalances.

7. **Multi-class (multi_class):**
   - Description:** Strategy to be used for multiclass problems.
   - **Options:** ‘auto’, ‘ovr’, ‘multinomial’
     - ‘auto’: Selects automatically according to the size of the data set.
     - ‘ovr’: One-vs-Rest (each class versus others).
     - ‘multinomial’: Direct multiclass target.
   - **Usage:** It is used in multi-class classification problems.

**Summary**

- Penalty (penalty):** Regularisation type (‘l1’, ‘l2’, ‘elasticnet’, ‘none’).
- Inverse of Regularisation Strength (C):** Regularisation strength (small values strong regularisation).
- **Solver (solver):** Optimisation algorithm (‘newton-cg’, ‘lbfgs’, ‘liblinear’, ‘sag’, ‘saga’).
- **Maximum Number of Iterations (max_iter):** Maximum number of iterations of the optimisation.
- **Class Weight (class_weight):** Class weights (‘balanced’, {dict}, None).
- **Multi-class (multi_class):** Strategy for multi-class problems (‘auto’, ‘ovr’, ‘multinomial’).

```python
from sklearn.model_selection import GridSearchCV

operations = [("scaler", StandardScaler()), ("logistic", LogisticRegression(max_iter=5000))]
model = Pipeline(steps=operations)
```

```python
# l1: Lasso, l2: Ridge
penalty = ["l1", "l2"]

# To obtain 20 C values between -1 and 5
# As C increases, regularisation decreases, vice versa The reason for using np.logspace for # C is that the optimal value of the regularisation is usually not within a certain range.
# Therefore, to try different values over a wide range and find the value that works best, we use a logarithmic scale
# It makes sense to create values. C = np.logspace(-1, 5, 20)

# solver ın dört değeri
# solver = ["lbfgs", "liblinear", "sag", "saga"]

# penalty, C, ve solver'ın tüm kombinasyonlarını elde etmek için
param_grid = [
    {
        "logistic__penalty": ["l1"],
        "logistic__C": C,
        "logistic__solver": ["liblinear", "saga"]
    },
    {
        "logistic__penalty": ["l2"],
        "logistic__C": C,
        "logistic__solver": ["newton-cg", "lbfgs", "liblinear", "sag", "saga"]
    },
    {
        "logistic__penalty": ["elasticnet"],
        "logistic__C": C,
        "logistic__solver": ["saga"],
        "logistic__l1_ratio": np.linspace(0, 1, 10)  # ElasticNet için l1_ratio ekliyoruz
    },
    {
        "logistic__penalty": ["none"],
        "logistic__solver": ["newton-cg", "lbfgs", "sag", "saga"]
    }
]
```

```python
grid_model = GridSearchCV(model, param_grid = param_grid, cv = 5, return_train_score = True)
```
